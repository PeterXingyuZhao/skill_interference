```python
import json
import math
from pathlib import Path

# Change this path if your JSON file is elsewhere
json_path = Path("/content/drive/MyDrive/subliminal-learning/results/rare_token_contains_ge2_5_probability_metric_MI_check.json")

with open(json_path, "r") as f:
    data = json.load(f)

per_query = data["mutual_information"]["per_query"]

def x_log_x_over_y(x: float, y: float) -> float:
    """
    Return x * log(x / y), with the convention 0 * log(0 / y) = 0.
    Logs are natural logs, so MI is in nats.
    """
    if x == 0:
        return 0.0
    if y == 0:
        raise ValueError("Encountered y=0 while x>0, invalid probability mixture.")
    return x * math.log(x / y)

def binary_mi_for_one_s(p_z1_c1: float, p_z1_c0: float) -> float:
    """
    Compute I(C; Z | S=s_i), assuming P(C=1)=P(C=0)=1/2.

    p_z1_c1 = P(Z=1 | C=1, s_i)
    p_z1_c0 = P(Z=1 | C=0, s_i)
    """
    p_z0_c1 = 1.0 - p_z1_c1
    p_z0_c0 = 1.0 - p_z1_c0

    # P(Z=z | s_i) = 1/2 P(Z=z | C=1, s_i) + 1/2 P(Z=z | C=0, s_i)
    p_z1 = 0.5 * p_z1_c1 + 0.5 * p_z1_c0
    p_z0 = 0.5 * p_z0_c1 + 0.5 * p_z0_c0

    mi = 0.5 * (
        x_log_x_over_y(p_z1_c1, p_z1)
        + x_log_x_over_y(p_z0_c1, p_z0)
    ) + 0.5 * (
        x_log_x_over_y(p_z1_c0, p_z1)
        + x_log_x_over_y(p_z0_c0, p_z0)
    )

    return mi

p_z1_given_c1_values = []
p_z1_given_c0_values = []
mi_per_query = []

for row in per_query:
    p1 = row["P_Z_eq_1_given_C_eq_1_S"]
    p0 = row["P_Z_eq_1_given_C_eq_0_S"]

    if not (0.0 <= p1 <= 1.0):
        raise ValueError(f"Invalid probability for C=1: {p1}")
    if not (0.0 <= p0 <= 1.0):
        raise ValueError(f"Invalid probability for C=0: {p0}")

    p_z1_given_c1_values.append(p1)
    p_z1_given_c0_values.append(p0)

    mi_i = binary_mi_for_one_s(p1, p0)
    mi_per_query.append(mi_i)

# I(C; Z | S) = average over s_i
I_C_Z_given_S = sum(mi_per_query) / len(mi_per_query)

# P(Z=1 | C=1) and P(Z=1 | C=0), averaged over S
P_Z1_given_C1 = sum(p_z1_given_c1_values) / len(p_z1_given_c1_values)
P_Z1_given_C0 = sum(p_z1_given_c0_values) / len(p_z1_given_c0_values)

ratio = (
    P_Z1_given_C1 / P_Z1_given_C0
    if P_Z1_given_C0 != 0
    else float("inf")
)

print(f"Number of queries: {len(per_query)}")
print(f"I(C; Z | S) in nats = {I_C_Z_given_S:.12g}")
print(f"I(C; Z | S) in bits = {I_C_Z_given_S / math.log(2):.12g}")
print(f"P(Z=1 | C=1) = {P_Z1_given_C1:.12g}")
print(f"P(Z=1 | C=0) = {P_Z1_given_C0:.12g}")
print(f"P(Z=1 | C=1) / P(Z=1 | C=0) = {ratio:.12g}")
```

### behavior & dataset context vector

```python
!python scripts/print_behavior_dataset_cosine_matrix.py --layer 26
```

```python
import json
from itertools import combinations
from pathlib import Path

import numpy as np
from loguru import logger

ROOT = Path("results/in-context_vector")
VECTOR_DIR = ROOT / "teacher_vectors_no_system_owl_999"
SWEEP_PATH = (
    VECTOR_DIR
    / "behavior_sampled_prompt_all_layers_23_24_26_alpha4.json"
)

TRAITS = ("eagle", "panda", "giraffe", "frog")


def cosine(a: np.ndarray, b: np.ndarray) -> float:
    """Compute signed cosine similarity between two hidden_dim vectors."""
    a = np.asarray(a, dtype=np.float64).reshape(-1)
    b = np.asarray(b, dtype=np.float64).reshape(-1)

    if a.shape != b.shape:
        raise ValueError(f"Shape mismatch: {a.shape} != {b.shape}")

    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))


def load_dataset_vector_at_layer(trait: str, layer: int) -> np.ndarray:
    """Load one dataset concept vector row for a model layer."""
    vector_path = VECTOR_DIR / f"{trait}_v_teacher_by_layer.npy"
    metadata_path = VECTOR_DIR / f"{trait}_v_teacher.metadata.json"

    vector_by_layer = np.load(vector_path)
    metadata = json.loads(metadata_path.read_text())

    row_by_layer = {
        int(model_layer): row
        for row, model_layer in enumerate(metadata["model_layer_indices"])
    }

    if layer not in row_by_layer:
        raise ValueError(
            f"{trait} dataset vector does not contain layer {layer}. "
            f"Available layers: {sorted(row_by_layer)}"
        )

    return vector_by_layer[row_by_layer[layer]]


sweep = json.loads(SWEEP_PATH.read_text())

best_layer_by_trait = {
    result["trait"]: int(result["best_result_by_delta"]["model_layer_index"])
    for result in sweep["trait_results"]
}

results = {
    "sweep_path": str(SWEEP_PATH),
    "best_layer_by_trait": best_layer_by_trait,
    "behavior_vs_dataset_best_layer": {},
    "dataset_vs_dataset_best_layer": {},
    "dataset_vs_dataset_same_layer": {},
}

dataset_best_vectors = {}

for trait in TRAITS:
    best_layer = best_layer_by_trait[trait]

    behavior_path = ROOT / f"trained_{trait}_vector.npy"
    behavior_vector = np.load(behavior_path)

    dataset_vector = load_dataset_vector_at_layer(trait, best_layer)
    dataset_best_vectors[trait] = dataset_vector

    value = cosine(behavior_vector, dataset_vector)

    results["behavior_vs_dataset_best_layer"][trait] = {
        "behavior_vector_path": str(behavior_path),
        "dataset_vector_dir": str(VECTOR_DIR),
        "best_layer": best_layer,
        "cosine": value,
    }

    logger.info(
        "{} behavior vs dataset L{} cosine = {:+.8f}",
        trait,
        best_layer,
        value,
    )


# Pairwise cosine between each trait's best-layer dataset vector.
# Example: panda L26 vs frog L23 if those are their respective best layers.
for trait_a, trait_b in combinations(TRAITS, 2):
    layer_a = best_layer_by_trait[trait_a]
    layer_b = best_layer_by_trait[trait_b]

    value = cosine(
        dataset_best_vectors[trait_a],
        dataset_best_vectors[trait_b],
    )

    pair_key = f"{trait_a}_L{layer_a}_vs_{trait_b}_L{layer_b}"
    results["dataset_vs_dataset_best_layer"][pair_key] = value

    logger.info(
        "dataset {} L{} vs {} L{} cosine = {:+.8f}",
        trait_a,
        layer_a,
        trait_b,
        layer_b,
        value,
    )


# Same-layer dataset-vector controls.
# This is usually the cleaner dataset-vs-dataset comparison.
for layer in sorted({best_layer_by_trait[trait] for trait in TRAITS}):
    layer_results = {}

    for trait_a, trait_b in combinations(TRAITS, 2):
        vector_a = load_dataset_vector_at_layer(trait_a, layer)
        vector_b = load_dataset_vector_at_layer(trait_b, layer)

        pair_key = f"{trait_a}_vs_{trait_b}"
        layer_results[pair_key] = cosine(vector_a, vector_b)

        logger.info(
            "dataset same layer L{}: {} vs {} cosine = {:+.8f}",
            layer,
            trait_a,
            trait_b,
            layer_results[pair_key],
        )

    results["dataset_vs_dataset_same_layer"][str(layer)] = layer_results


output_path = ROOT / "best_layer_behavior_dataset_alignment.json"
output_path.write_text(json.dumps(results, indent=2) + "\n")
logger.success("Saved results to {}", output_path)
```

```python
import json
from itertools import combinations
from pathlib import Path

import numpy as np
from loguru import logger


ROOT = Path("results/in-context_vector")
VECTOR_DIR = ROOT / "teacher_vectors_no_system_owl_999"
SWEEP_PATH = VECTOR_DIR / "behavior_sampled_prompt_all_layers_23_24_26_alpha4.json"
OUTPUT_PATH = ROOT / "best_layer_behavior_dataset_alignment_with_behavior_pairs.json"

TRAITS = ("eagle", "panda", "giraffe", "frog")


def cosine(a: np.ndarray, b: np.ndarray) -> float:
    """Compute signed cosine similarity between two hidden_dim vectors."""
    a = np.asarray(a, dtype=np.float64).reshape(-1)
    b = np.asarray(b, dtype=np.float64).reshape(-1)

    if a.shape != b.shape:
        raise ValueError(f"Shape mismatch: {a.shape} != {b.shape}")

    a_norm = float(np.linalg.norm(a))
    b_norm = float(np.linalg.norm(b))
    if a_norm == 0.0 or b_norm == 0.0:
        raise ValueError("Cannot compute cosine with a zero-norm vector.")

    return float(np.dot(a, b) / (a_norm * b_norm))


def load_dataset_vector_at_layer(trait: str, layer: int) -> np.ndarray:
    """Load one dataset concept vector row for a model layer."""
    vector_path = VECTOR_DIR / f"{trait}_v_teacher_by_layer.npy"
    metadata_path = VECTOR_DIR / f"{trait}_v_teacher.metadata.json"

    vector_by_layer = np.load(vector_path)
    metadata = json.loads(metadata_path.read_text())

    row_by_layer = {
        int(model_layer): row
        for row, model_layer in enumerate(metadata["model_layer_indices"])
    }

    if layer not in row_by_layer:
        raise ValueError(
            f"{trait} dataset vector does not contain layer {layer}. "
            f"Available layers: {sorted(row_by_layer)}"
        )

    return np.asarray(vector_by_layer[row_by_layer[layer]], dtype=np.float64).reshape(-1)


def center_vectors(vectors: dict[str, np.ndarray]) -> dict[str, np.ndarray]:
    """Center trait vectors by subtracting the across-trait mean vector."""
    matrix = np.stack(
        [np.asarray(vectors[trait], dtype=np.float64).reshape(-1) for trait in TRAITS],
        axis=0,
    )
    centered = matrix - matrix.mean(axis=0, keepdims=True)
    return {
        trait: centered[index]
        for index, trait in enumerate(TRAITS)
    }


def pairwise_cosines(vectors: dict[str, np.ndarray]) -> dict[str, float]:
    """Compute pairwise cosines between trait vectors."""
    out = {}
    for trait_a, trait_b in combinations(TRAITS, 2):
        key = f"{trait_a}_vs_{trait_b}"
        out[key] = cosine(vectors[trait_a], vectors[trait_b])
    return out


sweep = json.loads(SWEEP_PATH.read_text())

best_layer_by_trait = {
    result["trait"]: int(result["best_result_by_delta"]["model_layer_index"])
    for result in sweep["trait_results"]
}

results = {
    "sweep_path": str(SWEEP_PATH),
    "best_layer_by_trait": best_layer_by_trait,
    "behavior_vs_dataset_best_layer": {},
    "behavior_vs_behavior": {},
    "centered_behavior_vs_behavior": {},
    "dataset_vs_dataset_best_layer": {},
    "dataset_vs_dataset_same_layer": {},
}

behavior_vectors = {}
dataset_best_vectors = {}

for trait in TRAITS:
    best_layer = best_layer_by_trait[trait]

    behavior_path = ROOT / f"trained_{trait}_vector.npy"
    behavior_vector = np.asarray(np.load(behavior_path), dtype=np.float64).reshape(-1)
    behavior_vectors[trait] = behavior_vector

    dataset_vector = load_dataset_vector_at_layer(trait, best_layer)
    dataset_best_vectors[trait] = dataset_vector

    value = cosine(behavior_vector, dataset_vector)

    results["behavior_vs_dataset_best_layer"][trait] = {
        "behavior_vector_path": str(behavior_path),
        "dataset_vector_dir": str(VECTOR_DIR),
        "best_layer": best_layer,
        "cosine": value,
    }

    logger.info(
        "{} behavior vs dataset L{} cosine = {:+.8f}",
        trait,
        best_layer,
        value,
    )


# Pairwise cosine between behavior concept vectors.
results["behavior_vs_behavior"] = pairwise_cosines(behavior_vectors)

for pair_key, value in results["behavior_vs_behavior"].items():
    logger.info("behavior {} cosine = {:+.8f}", pair_key, value)


# Pairwise cosine between centered behavior concept vectors.
centered_behavior_vectors = center_vectors(behavior_vectors)
results["centered_behavior_vs_behavior"] = pairwise_cosines(centered_behavior_vectors)

for pair_key, value in results["centered_behavior_vs_behavior"].items():
    logger.info("centered behavior {} cosine = {:+.8f}", pair_key, value)


# Pairwise cosine between each trait's best-layer dataset vector.
for trait_a, trait_b in combinations(TRAITS, 2):
    layer_a = best_layer_by_trait[trait_a]
    layer_b = best_layer_by_trait[trait_b]

    value = cosine(
        dataset_best_vectors[trait_a],
        dataset_best_vectors[trait_b],
    )

    pair_key = f"{trait_a}_L{layer_a}_vs_{trait_b}_L{layer_b}"
    results["dataset_vs_dataset_best_layer"][pair_key] = value

    logger.info(
        "dataset {} L{} vs {} L{} cosine = {:+.8f}",
        trait_a,
        layer_a,
        trait_b,
        layer_b,
        value,
    )


# Same-layer dataset-vector controls.
for layer in sorted({best_layer_by_trait[trait] for trait in TRAITS}):
    layer_results = {}

    for trait_a, trait_b in combinations(TRAITS, 2):
        vector_a = load_dataset_vector_at_layer(trait_a, layer)
        vector_b = load_dataset_vector_at_layer(trait_b, layer)

        pair_key = f"{trait_a}_vs_{trait_b}"
        layer_results[pair_key] = cosine(vector_a, vector_b)

        logger.info(
            "dataset same layer L{}: {} vs {} cosine = {:+.8f}",
            layer,
            trait_a,
            trait_b,
            layer_results[pair_key],
        )

    results["dataset_vs_dataset_same_layer"][str(layer)] = layer_results


OUTPUT_PATH.write_text(json.dumps(results, indent=2) + "\n")
logger.success("Saved results to {}", OUTPUT_PATH)

def centered_pairwise_cosines_for_dataset_layer(layer: int) -> dict[str, float]:
    vectors = {
        trait: load_dataset_vector_at_layer(trait, layer)
        for trait in TRAITS
    }

    matrix = np.stack([vectors[trait] for trait in TRAITS], axis=0)
    centered = matrix - matrix.mean(axis=0, keepdims=True)

    centered_vectors = {
        trait: centered[index]
        for index, trait in enumerate(TRAITS)
    }

    return pairwise_cosines(centered_vectors)


results["centered_dataset_vs_dataset_same_layer"] = {}

for layer in sorted({best_layer_by_trait[trait] for trait in TRAITS}):
    layer_results = centered_pairwise_cosines_for_dataset_layer(layer)
    results["centered_dataset_vs_dataset_same_layer"][str(layer)] = layer_results

    for pair_key, value in layer_results.items():
        logger.info(
            "centered dataset same layer L{}: {} cosine = {:+.8f}",
            layer,
            pair_key,
            value,
        )
```
